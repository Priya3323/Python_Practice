```python
x=int(input("Enter the number"))
x=str(x)
print(len(x))
list1=[]
for i in x:
    list1.append(i)
print(list1)

flag=False
for i in range(len(list1)):
    if list1[i] == list1[-1-i]:
        flag=False
    else:
        flag=True
        break
if flag==False:
    print("Number is palindrome")
else:
        print("Number is not palindrome")
```

```python
# Imagine you are building an online shopping app like Amazon. When a customer reaches the checkout page, they need to click a "Pay" button.
#The app doesn't care how you pay; it just wants to execute a payment.
# Here is how polymorphism handles this seamlessly in code:
```

```python
# Form 1: Credit Card Payment
class CreditCard:
    def process_payment(self, amount):
        return f"Charging ₹{amount} to your Visa card. (Authorizing bank...)"

# Form 2: UPI / UPI App Payment
class UPI:
    def process_payment(self, amount):
        return f"Requesting ₹{amount} via UPI. (Opening your phone app...)"

# Form 3: Cash on Delivery
class CashOnDelivery:
    def process_payment(self, amount):
        return f"Order placed for ₹{amount}. Collect cash at the doorstep."


# The Poly-morphic Function (The "Pay Button")
# It takes ANY payment object and simply calls 'process_payment'
def click_pay_button(payment_method, bill_amount):
    print(payment_method.process_payment(bill_amount))
    
# Real-time execution based on user selection:
user_choice_1 = CreditCard()
user_choice_2 = UPI()


click_pay_button(user_choice_1, 1500)  # Output: Charging ₹1500 to your Visa card...
click_pay_button(user_choice_2, 450)   # Output: Requesting ₹450 via UPI...
```

```python
# Method Overriding---Method Overriding is when a child class redefines a method that it inherited from its parent class.
# Parent Class
class Parent:
    def house_rules(self):
        print("Curfew is at 8:00 PM.")

# Child Class inherits from Parent
class Child(Parent):
    # Method Overriding: Changing the rule for the child!
    def house_rules(self):
        print("Curfew is at 11:00 PM.")

# Execution
someone = Child()
someone.house_rules()  # Output: Curfew is at 11:00 PM.
```

# Abstraction in OOP

```python
# Abstract Class (In Python): A blueprint class that cannot be used to create objects directly. It is meant to be inherited by child classes.
```

```python
# Abstract Method (unset @abstractmethod ): A mandatory rule declared in the parent class. 
# Every child class must write its own version of this method to be allowed to run.
```

```python
# In Python, the @ symbol is called a decorator.
```

```python
from abc import ABC, abstractmethod


class company(ABC): # Abstract - Parent class
# --------------Abstract Method---------------
    @abstractmethod
    def info(self): # Abstract Method 1
        print("Information")
    def details(self): # Method 2
        print("Details")
#--------------------------
class dept_HR(company): # child class
    def KPI(self):
        print("KPI")
    def info(self):
        pass
    
#-------------------------
IT_1001=dept_HR() # object creation
IT_1001.KPI() # object cannot use the class unless "info" method is implemented in child class
```

```python
# 🚗 Real-Life Example: A Coffee Machine

# Think of a smart coffee machine:
# • The Interface (What you see): A single button that says "Make Coffee".
# • The Hidden Complexity (What you don't see): The machine boiling water, grinding beans, pumping pressure, and mixing the milk.
# You don't need to know how it heats the water to get your caffeine fix; you just press the button.
```

```python
from abc import ABC, abstractmethod

# 1. The Abstract Blueprint (The Simple Interface)
class CoffeeMachine(ABC):
    
    @abstractmethod
    def make_coffee(self):
        """This is a rule: Every coffee machine MUST have a make_coffee button"""
        pass

# 2. The Real Implementation (The Hidden Complexity)
class PremiumEspressoMachine(CoffeeMachine):
    
    def make_coffee(self):
        # All the messy, complex steps are hidden inside this method
        self._grind_beans()
        self._boil_water()
        self._brew()
        print("☕ Your premium Espresso is ready!")

    def _grind_beans(self):
        print("Grinding coffee beans at 1500 RPM...")

    def _boil_water(self):
        print("Heating water to exactly 92°C...")

    def _brew(self):
        print("Extracting coffee under 9 bars of pressure...")

# 3. How the User Interacts with it
my_machine = PremiumEspressoMachine()

# The user only calls this ONE simple method
my_machine.make_coffee()
```

```python

# Bracket Expressions (Character Sets)

• [abc] : Matches any single character that is either a, b, or c.
• [^abc] : Matches any character except a, b, or c (the ^ inside brackets means "not").
• [a-z] : Matches any lowercase letter from a to z.
• [A-Z] : Matches any uppercase letter from A to Z.
• [0-9] : Matches any single digit from 0 to 9.
• [a-zA-Z0-9] : Matches any alphanumeric character (any letter or digit).
```

```python
# Special Shortcuts (Meta-characters)

• . : Matches any single character except a new line (a wildcard).
• \d : Matches any digit (same as [0-9]).
• \D : Matches any non-digit (same as [^0-9]).
• \w : Matches any word character (letters, numbers, and underscores).
• \s : Matches any whitespace (spaces, tabs, new lines).
```

```python
# Anchors (Where to Look)

• ^ : Matches the start of a string or line (when used outside of brackets).
• $ : Matches the end of a string or line.
• \b : Matches a word boundary (the start or end of a whole word).
```

```python
# 🛠️ Logic & Grouping

• | : Acts like an OR operator (e.g., cat|dog matches "cat" or "dog").
• () : Groups tokens together and captures the text inside them.
```

```python
# 🐢 Standard Quantifiers (Greedy)

By default, these are greedy, meaning they will match as much text as they possibly can.
• * : Matches 0 or more times.
• + : Matches 1 or more times.
• ? : Matches 0 or 1 time (makes the item optional).
• {n} : Matches exactly n times.
• {n,} : Matches n or more times.
• {n,m} : Matches between n and m times (inclusive).
```

```python
# Lazy Quantifiers (Non-Greedy)

If you add a ? after any standard quantifier, it becomes lazy. This tells regex to match the smallest amount of text possible.
• *? : Matches 0 or more times (stops at the very first match).
• +? : Matches 1 or more times (stops at the very first match).
• ?? : Matches 0 or 1 time (prefers 0 times if possible).
• {n,}? : Matches n or more times (stops exactly at n).
• {n,m}? : Matches between n and m times (stops exactly at n).
```

# Concatenate arrays

```python
a=np.arange(1,19).reshape(6,3)
b=np.arange(7,13).reshape(6,1)
c=np.arange(1,4).reshape(1,3)
print(a)
print("-------------------")
print(b)
print("-------------------")
print(c)
```

```python
# cancatenate ((arrr1,arr2))
# it is used to combine two arrays and it take them in a tuple
np.concatenate((a,b),axis=1) # Horizontal concatenate
```

```python
# horizontal cn=oncatenate ----rows should be same
# vertical concatenate----columns should be same
```

```python
d=np.arange(1,19).reshape(6,3)
e=np.arange(7,13).reshape(2,3)

np.concatenate((d,e),axis=0) # vertical concatenate--- default --- but we can write its axis as zero
```
