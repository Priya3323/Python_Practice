```python
x=int(input("Enter the number"))
x=str(x)
print(len(x))
list1=[]
for i in x:
    list1.append(i)
print(list1)

flag=False
for i in range(len(list1)):
    if list1[i] == list1[-1-i]:
        flag=False
    else:
        flag=True
        break
if flag==False:
    print("Number is palindrome")
else:
        print("Number is not palindrome")
```

# Exception Handling

```python
try:
    x=int(input("Enter dividend:"))
    y=int(input("Enter divisor:"))
    z=x/y
except:
    print("Enter numbers only and divisor should not be zero")

print(z)
```

```python
# 2nd approach
try:
    x=int(input("Enter dividend:"))
    y=int(input("Enter divisor:"))
    z=x/y
except BaseException as E:
    print(E)

print(z)
```

```python
try:
    x=int(input("Enter dividend:"))
    y=int(input("Enter divisor:"))
    z=x/y
except ZeroDivisionError as ZE:
    print(ZE)
except ValueError as VE:
    print(VE)
else:
    print(z)
finally:
    print("Program ends here, Thank You!")
```
