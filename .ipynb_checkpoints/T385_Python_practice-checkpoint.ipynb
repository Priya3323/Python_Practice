# Break, continue and Pass, we can use these parameters only in loops

```python
# to print value from 1 to 10 but stop when i=7
i=1
while i<=10:
    if i==7:
        break
    else:
        print(i)
    i=i+1
print("Program Ends")
```

```python
# write program to find the prime numbers
num=int(input("Enter Number"))
divisor=num-1
flag = 1
while divisor>=2:
    if num%divisor==0:
        flag = 0
        break
    else:
        pass
    divisor=divisor-1
if flag == 1:
    print("Number is Prime")
else:
    print("Not a prime number")
```

```python
# write program to find the factorial of any number
n=int(input("Enter Number"))
fact = 1
for i in range (1, n+1):
    fact = fact*i
print("Factorial = ", fact)
```

# WAP to check wheather the number is palimdrom or not (palimdrome means a number start with same no and end with same number/ mirror image, eg 343, 999,45454 (even number count is not palimdrom)


```python
x=int(input("Enter the number"))
x=str(x)
print(len(x))
list1=[]
for i in x:
    list1.append(i)
print(list1)

flag=False
for i in range(len(list1)):
    if list1[i] == list1[-1-i]:
        flag=False
    else:
        flag=True
        break
if flag==False:
    print("Number is palindrome")
else:
        print("Number is not palindrome")
```
